```python
# Your code here
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Notebook is ready!")
```

```python
# === CONFIG ===
input_file = "BDO_results_rdkit.csv"  # input file
output_file = "BDO_results_rdkit_binary.csv"  # output file
prob_column = "ACTIVITY"  # column with probabilities
threshold = 0.62  # decision threshold

# === LOAD DATA ===
df = pd.read_csv(input_file)

# === CHECK COLUMN EXISTS ===
if prob_column not in df.columns:
    raise ValueError(f"Column '{prob_column}' not found in CSV. Available columns: {list(df.columns)}")

# === CREATE BINARY COLUMN BASED ON THRESHOLD ===
df["activity_growth"] = (df[prob_column] >= threshold).astype(int)

```

```python
# === SAVE TO NEW CSV ===
df.to_csv(output_file, index=False)

print(f"✅ Processed file saved to '{output_file}'")
print(f"Applied threshold: {threshold}")
print(df.head())
```
